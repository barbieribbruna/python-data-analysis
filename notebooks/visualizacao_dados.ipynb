```python
# IMPORTAÇÕES

# Importa a biblioteca Pandas para manipulação e análise de dados
import pandas as pd

# Importa o módulo pyplot da biblioteca Matplotlib para criação de gráficos
import matplotlib.pyplot as plt

# Importa a biblioteca Seaborn para visualizações estatísticas
import seaborn as sns

# Importa o módulo 'Express' da biblioteca Plotly para criação de gráficos interativos
import plotly.express as px
```

```python
# LEITURA DA BASE DE DADOS

# Lê a base tratada para análise visual
dados = pd.read_csv('../data/dados-tratados.csv')
```

```python
dados.head()
```

```python
# Conta a quantidade de rgistros de cada gênero
genero_counts = dados['GENERO'].value_counts()
```

```python
# Cria uma nova figura para o gráfico
plt.figure()

# Cria gráfico de barras com os gêneros no eixo X e suas contagens no eixo Y
plt.bar(genero_counts.index, genero_counts.values)

# Define o título do gráfico
plt.title('Quantidade de pessoas por gênero na área de Dados')

# Define o nome do eixo X
plt.xlabel('Gênero')

# Define o nome do eixo Y
plt.ylabel('Contagem')

# Exibe o gráfico
plt.show()
```

```python
# Cria uma nova figura para o gráfico
plt.figure()

# Cria um gráfico de contagem da coluna 'GENERO', usando a paleta de cores pastel
sns.countplot(data = dados, x = 'GENERO', palette = 'pastel')

# Define o título do gráfico
plt.title('Quantidade de pessoas por gênero na área de Dados')

# Define o nome do eixo X
plt.xlabel('Gênero')

# Define o nome do eixo Y
plt.ylabel('Contagem')

# Exibe o gráfico
plt.show()
```

```python
# Agrupa os dados por idade e calcula a média salarial de cada faixa etária
salario_por_idade = dados.groupby('IDADE')['SALARIO'].mean()
```

```python
salario_por_idade
```

```python
# Cria uma figura para o gráfico
plt.figure()

# Cria um gráfico de linha da média salarial por idade | 'marker' adiciona marcadores nos pontos e 'linestyle' define o tipo de linha
plt.plot(salario_por_idade.index, salario_por_idade.values, marker='o', linestyle='--')

# Define o nome do eixo X
plt.xlabel('Idade')

# Define o nome do eixo Y
plt.ylabel('Média de salário')

# Define o título do gráfico
plt.title('Média de salário por idade')

# Adiciona grade ao gráfico
plt.grid(True)

# Exibe o gráfico
plt.show()
```

```python
# Cria um gráfico de linha interativo com Plotly
# reset_index() transforma o índice em coluna novamente
fig = px.line(salario_por_idade.reset_index(), x='IDADE', y='SALARIO', title='Média de salário por idade', markers=True)

# Exibe o gráfico interativo
fig.show()
```

```python
# Cria uma nova figura com tamanho personalizado
plt.figure(figsize=(10,5))

# Cria um gráfico de dispersão entre idade e salário
plt.scatter(dados['IDADE'], dados['SALARIO'], alpha=0.5)

# Define o nome do eixo X
plt.xlabel('Idade')

# Define o nome do eixo Y
plt.ylabel('Salário')

# Define o título do gráfico
plt.title('Dispersão de salário por idade')

# Adiciona grade ao gráfico
plt.grid(True)

# Exibe o gráfico
plt.show()
```

```python
# Cria um gráfico interativo de dispersão com Plotly
fig = px.scatter(dados, x='IDADE', y='SALARIO', title='Dispersão de salário por idade')

# Exibe o gráfico interativo
fig.show()
```
